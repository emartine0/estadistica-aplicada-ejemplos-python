# Buscando Relaciones entre Variables

## Correlacion de Pearson

### Procedimiento
- Importar librerias
- Cargar la primera hoja de trabajo de excel en Pandas
- Prueba de Normalidad (SHAPIRO-WILK)
- Prueba de Linelidad (VISUAL)
- Prueba correlacion de Pearson

```python
#__ Cargar librerias
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import shapiro, pearsonr
```

```python
#___ Cargar la hoja de trabajo en un dataframe
corr_df = pd.read_excel('datos_analisis_estadistico.xlsx', sheet_name = 'CORRELACIÓN')
corr_df
```

```python
#___ Prueba de normalidad (ANALITICA SHAPIRO-WILK)
###___ La funcion sp.shapiro() verifica que una o mas columnas en el dataframe siguen una distribucion normal.
prueba_norm_peso = shapiro(corr_df['PESO'])
prueba_norm_altura = shapiro(corr_df['ALTURA'])
print(prueba_norm_peso, prueba_norm_altura)
```

```python
#___ Prueba de linealidad VISUAL
corr_df.plot.scatter(x='ALTURA', y='PESO', figsize=(4,3))
plt.xlabel('ALTURA', fontsize=18)
plt.ylabel('PESO', fontsize=18)
plt.show()
```

```python
#___ Prueba de correlacion de PEARSON METODO A
prueba_pearson = pearsonr(corr_df['ALTURA'], corr_df['PESO'])
prueba_pearson
```

```python
#___ Prueba de correlacion de PEARSON METODO B
corr_df.drop('ELEMENTO', axis=1).corr()
```
