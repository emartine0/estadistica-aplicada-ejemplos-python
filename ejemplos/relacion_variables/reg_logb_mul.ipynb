# Buscando Relación entre Variables

## Regresion logistica binaria multiple

### Procedimiento
- Importar librerias
- Cargar la hoja de trabajo de excel en Pandas
- Prueba de Multicolinealidad (VIF)
- Análisis de Regresion Logistica Binaria Multiple

```python
#__ Cargar librerias
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.api import Logit, add_constant
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
#___ Cargar la hoja de trabajo en un dataframe
reglogb_m_df = pd.read_excel('ejemplos.xlsx', sheet_name='REG LOG B M')
reglogb_m_df.columns = ['ELEMENTO', 'EDAD', 'SUENO', 'DIABETES']
reglogb_m_df
```

```python
#___ Visualizacion
fig, (ax1, ax2) = plt.subplots(1,2, figsize=(12,6))

ax1.scatter(x='EDAD', y='DIABETES', data=reglogb_m_df)
ax1.set_facecolor("#343837")          # Inner color
ax1.figure.set_facecolor("#343837")    # Outer color
ax1.set_xlabel('EDAD', color='tomato')
ax1.set_ylabel('DIABETES', color='tomato')
ax1.tick_params(labelcolor='white')

ax2.scatter(x='SUENO', y='DIABETES', data=reglogb_m_df)
ax2.set_facecolor("#343837")          # Inner color
ax2.figure.set_facecolor("#343837")    # Outer color
ax2.set_xlabel('SUENO', color='tomato')
ax2.set_ylabel('DIABETES', color='tomato')
ax2.tick_params(labelcolor='white')

plt.show()
```

```python
#___ Prueba analitica de multicolinealidad
###___ Interpretacion del factor de inflacion de la varianza (VIF)
###___ Valores cercanos a 1 i indican que los predictores son independientes
###___ Valores entre 1 y 5 muestran una correlacion moderada que es en ocasiones aceptable
###___ Valores arriba de 10 señalan que hay una multicolinealidad problematica
datos_vif = pd.DataFrame()
datos_vif["Predictor"] = ['EDAD', 'SUENO']
datos_vif["VIF"] = [variance_inflation_factor(reglogb_m_df[['EDAD', 'SUENO']], i) for i in range(reglogb_m_df[['EDAD', 'SUENO']].values.shape[1])]
datos_vif
```

```python
#___ Análisis de regresion logistica binaria simple
y = reglogb_m_df['DIABETES']
X = reglogb_m_df[['EDAD', 'SUENO']]
###___ Agregar constante a las variables independientes (interseccion)
X_const = add_constant(X)
###___ Crear el modelo de regresion logistica binaria multiple
modelo = Logit(y, X_const)
###___ Ajustar el modelo
resultado = modelo.fit()
###___ Resultados
resultado.summary()
```
