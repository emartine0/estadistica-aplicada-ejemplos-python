# Buscando Relación entre Variables

## Regresion logistica binaria simple

### Procedimiento
- Importar librerias
- Cargar la hoja de trabajo de excel en Pandas
- Análisis de Regresion Logistica Binaria Simple

```python
#__ Cargar librerias
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.api import Logit, add_constant
```

```python
#___ Cargar la hoja de trabajo en un dataframe
reglogb_s_df = pd.read_excel('ejemplos.xlsx', sheet_name='REG LOG B S')
reglogb_s_df.columns = ['ELEMENTO', 'EDAD', 'DIABETES']
reglogb_s_df
```

```python
#___ Visualizacion
fig, ax1 = plt.subplots(figsize=(8,4))

ax1.scatter(x='EDAD', y='DIABETES', data=reglogb_s_df)
ax1.set_facecolor("#343837")          # Inner color
ax1.figure.set_facecolor("#343837")    # Outer color
ax1.set_xlabel('EDAD', color='tomato')
ax1.set_ylabel('DIABETES', color='tomato')
ax1.tick_params(labelcolor='white')

plt.show()
```

```python
#___ Análisis de regresion logistica binaria simple
y = reglogb_s_df['DIABETES']
X = reglogb_s_df['EDAD']
## Add a constant (intercept) to the independent variable
X_const = add_constant(X)
# Create the Logit model
modelo = Logit(y, X_const)
# Fit the model
resultado = modelo.fit()
# Print the summary of the regression results
resultado.summary()
```
