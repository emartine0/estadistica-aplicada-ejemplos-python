# Buscando Relación entre Variables

## Regresion lineal simple

### Procedimiento
- Importar librerias
- Cargar la hoja de trabajo de excel en Pandas
- Prueba de Linelidad (VISUAL)
- Prueba de Normalidad (SHAPIRO-WILK)
- Prueba de Homocedasticidad (BREUCH-PAGAN)
- Prueba de Independencia de los Residuos (DRUBIN-WATSON)
- Prueba de Regresion Lineal Simple

```python
#__ Cargar librerias
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import shapiro, linregress
from statsmodels.api import add_constant, OLS, qqplot
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.stattools import durbin_watson
```

```python
#___ Cargar la hoja de trabajo en un dataframe
reglin_s_df = pd.read_excel('datos_analisis_estadistico.xlsx', sheet_name = 'REG LINEAL S')
reglin_s_df.columns = ['ELEMENTO', 'EDAD', 'IMC']
reglin_s_df
```

```python
#___ Prueba de linealidad VISUAL SEABORN
plt.figure(figsize=(12,4), dpi=200)
ax = sns.scatterplot(x='EDAD', y='IMC', data=reglin_s_df)

ax.set_facecolor("#343837")          # Inner color
ax.figure.set_facecolor("#343837")    # Outer color
ax.set_xlabel('EDAD', color='tomato')
ax.set_ylabel('IMC', color='tomato')
ax.tick_params(labelcolor='white')
plt.show()
```

```python
#___ Obtener los residuos para prueba de normalidad METODO A
pendiente, interseccion, r, p, es = linregress(reglin_s_df['EDAD'], reglin_s_df['IMC'])
recta_reg = pendiente * reglin_s_df['EDAD'] + interseccion
```

```python
#___ Prueba de normalidad sobre los residuos (SCIPY)
residuos_A = recta_reg - reglin_s_df['IMC']
prueba_norm_A = shapiro(residuos_A)
prueba_norm_A
```

```python
#___ Obtener los residuos para prueba de normalidad METODO B
X = add_constant(reglin_s_df['EDAD'])
y = reglin_s_df['IMC']

modelo = OLS(y, X).fit()
residuos_B = modelo.resid
```

```python
#___ Prueba de normalidad sobre los residuos (SCIPY)
prueba_norm_B = shapiro(residuos_B)
prueba_norm_B
```

```python
#___ Prueba de normalidad METODO VISUAL
qqplot(residuos_B, line='s')
plt.title("Grafica Q-Q del Modelo de Residuos")
plt.show()
```

```python
#___ Prueba de Homocedasticidad Breuch-Pagan (statsmodels)
prueba_homos = het_breuschpagan(modelo.resid, modelo.model.exog)
prueba_homos
###___ Output: (lagrange_multiplier_statistic (lms), p-value_lms, f-value, f-p-value)
```

```python
#___ Prueba de Homocedasticidad VISUAL (statsmodels)
#___ Prueba residuos independientes.
###___ Se obtiene la homocedasticidad si los puntos estan distribuidos aleatoria y igualitariamente alrededor y=0
valores_ajustados = modelo.fittedvalues

fig, ax = plt.subplots(figsize=(6, 3))
ax.scatter(valores_ajustados, residuos_B)
ax.set_facecolor("#343837")        
ax.figure.set_facecolor("#343837")   
ax.axhline(y=0, color='tomato', linestyle='--', linewidth=1.5)
ax.set_xlabel('Residuos', color='tomato')
ax.set_ylabel('Valores Ajustados', color='tomato')
ax.tick_params(labelcolor='white')
plt.show()
```

```python
#___ Prueba analitica residuos independientes (DURBIN-WATSON)
###___ Interpretacion del estadistico Durbin-Watson:
###___ Cerca de 2: No auto-correlacion.
###___ Cerca de 0: auto-correlacion positiva.
###___ Cerca de 4: auto-correlacion negativa.
prueba_res_indep = durbin_watson(residuos_B)
prueba_res_indep
```
