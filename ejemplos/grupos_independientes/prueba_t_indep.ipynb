# Buscando Diferencias entre Grupos Independientes

## Prueba T para grupos Independientes

### Procedimiento
- Cargar la primera hoja de trabajo de excel en Pandas
- Prueba de Normalidad (SHAPIRO-WILK)
- Prueba de homocedasticidad (LEVENE)
- Es necesario dividir el conjunto de datos en categorias (variables independientes cualitativas)
- Prueba T de Student

```python
#___ Cargar librerias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
```

```python
#___ Cargar la hoja de trabajo correspondiente como un dataframe
t_indep_df = pd.read_excel('datos_analisis_estadistico.xlsx', sheet_name='PRUEBA T INDEP')
t_indep_df
```

```python
#___ Examinar de manera visual los datos en la columna y limpiar 
t_indep_df.columns = ['ELEMENTO', 'SEXO', 'ALTURA']
plt.figure(figsize=(2, 3))
sns.barplot(data=t_indep_df, x='SEXO', y='ALTURA', estimator=np.mean, errorbar='sd', capsize=0.5)
plt.ylim(1,1.75)
```

```python
#___ Prueba de normalidad (SHAPIRO-WILK)
##___ Devuelve el estadistico de prueba y el valor p
prueba_norm = stats.shapiro(t_indep_df['ALTURA'])
prueba_norm
```

```python
#___ Prueba de homocedasticidad (LEVENE)
##___ Es necesario dividir el conjunto de datos de acuerdo a las categorias (Masculino-Femenino)
mas_df = t_indep_df[t_indep_df['SEXO'] == 'MASCULINO']
fem_df = t_indep_df[t_indep_df['SEXO'] == 'FEMENINO']
prueba_homos = stats.levene(mas_df['ALTURA'], fem_df['ALTURA'])
prueba_homos
```

```python
#___ Prueba T de Student
prueba_t_indep = stats.ttest_ind(mas_df['ALTURA'], fem_df['ALTURA'])
prueba_t_indep
```

```python
#___ Debido a que los datos estan normalmente distribuidos es apropiado utilizar una grafica de barras
promedio_sexo = t_indep_df[['SEXO','ALTURA']].groupby('SEXO').mean()
estd_sexo =  t_indep_df[['SEXO','ALTURA']].groupby('SEXO').std()

# NOTAR QUE promedio_sexo.index Y promedio_sexo['ALTURA] ESTAN ORDENADOS DE LA MISMA FORMA
plt.figure(figsize=(2, 3))  # Ancho de 10 pulgadas, alto de 5 pulgadas
plt.bar(promedio_sexo.index, promedio_sexo['ALTURA'], width=0.5)
plt.ylim(1,2)
plt.errorbar(promedio_sexo.index, promedio_sexo['ALTURA'], yerr=estd_sexo, fmt='o', color='r', capsize=20)
plt.xlabel('SEXO')
plt.ylabel('ALTURA (MTS)')
```
