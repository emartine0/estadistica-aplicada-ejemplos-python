# Buscando Diferencias entre Grupos Independientes

## ANOVA de 1 factor para grupos Independientes

### Procedimiento
- Importar librerias
- Cargar la primera hoja de trabajo de excel en Pandas
- Limpiar los datos
- Dividir el conjunto de datos en categorias (variables cualitativas independientes)
- Prueba de Normalidad (SHAPIRO-WILK)
- Prueba de homocedasticidad (LEVENE)
- Prueba ANOVA 1 Factor
- (opcional) Para la categoria con 3 o mas grupos HSD Tukey

```python
#__ Cargar librerias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as sps
```

```python
#___ Cargar la hoja de trabajo en un dataframe
anova_1_df = pd.read_excel('datos_analisis_estadistico.xlsx', sheet_name = 'ANOVA 1 FACTOR')
anova_1_df
```

```python
#___ Limpiar el dataframe
anova_1_df.columns = ['ELEMENTO', 'EJERCICIO', 'IMC']
anova_1_df = anova_1_df.drop('ELEMENTO', axis=1)
anova_1_df
```

```python
#___ Dividir los datos en categorias (variables cualitativas independientes)
anova_1_n_df = anova_1_df[anova_1_df['EJERCICIO'] == 'NUNCA']
anova_1_e_df = anova_1_df[anova_1_df['EJERCICIO'] == 'EVENTUAL']
anova_1_s_df = anova_1_df[anova_1_df['EJERCICIO'] == 'SIEMPRE']
```

```python
#___ Prueba de normalidad ANOVA (SHAPIRO-WILK)
prueba_norm = sps.shapiro(anova_1_df['IMC'])
prueba_norm
```

```python
#___ Prueba de homocedasticidad (LEVENE)
prueba_homos = sps.levene(anova_1_n_df['IMC'], anova_1_e_df['IMC'], anova_1_s_df['IMC'])
prueba_homos
```

```python
#___ Prueba ANOVA 1 Factor
prueba_anova_1 = sps.f_oneway(anova_1_n_df['IMC'], anova_1_e_df['IMC'], anova_1_s_df['IMC'])
prueba_anova_1
```

```python
#___ Prueba Tukey
prueba_tukey = sps.tukey_hsd(anova_1_n_df['IMC'], anova_1_e_df['IMC'], anova_1_s_df['IMC'])
print(prueba_tukey)
```

```python
#___ Grafica de error promedio
##___ Prepararr el dataframe
promedios = anova_1_df.groupby('EJERCICIO').mean()
destd = anova_1_df.groupby('EJERCICIO').std()

plt.figure(figsize=(3, 4))
plt.bar(promedios.index, promedios['IMC'], width=0.5)
plt.errorbar(promedios.index, promedios['IMC'], yerr=destd['IMC'], fmt='o', color='r', capsize=20)
plt.xlabel('TEMPORALIDAD')
plt.ylabel('IMC')
plt.title('ERROR PROMEDIO')
```
