# Buscando Diferencias entre Grupos Pareados

## Prueba T para grupos pareados

### Procedimiento
- Importar librerias
- Cargar la primera hoja de trabajo de excel en Pandas
- Limpiar los datos
- Prueba de Normalidad (SHAPIRO-WILK)
- Prueba T para grupos pareados

```python
#__ Cargar librerias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as sps
```

```python
#___ Cargar la hoja de trabajo en un dataframe
t_par_df = pd.read_excel('datos_analisis_estadistico.xlsx', sheet_name='PRUEBA T PAR')
t_par_df
```

```python
#___ Limpiar el dataframe
t_par_df.columns = ['PACIENTE', 'ANTES', 'DESPUES']
t_par_df = t_par_df.drop('PACIENTE', axis=1)
t_par_df
```

```python
#___ La prueba de normalidad (SHAPIRO-WILK) se corre sobre la diferencia de las dos muestras
t_par_df['DIFERENCIA'] = t_par_df['DESPUES'] - t_par_df['ANTES']
```

```python
#___ Prueba de normalidad ANOVA (SHAPIRO-WILK)
prueba_norm = sps.shapiro(t_par_df['DIFERENCIA'])
prueba_norm
```

```python
#___ Prueba T para grupos pareados (con la opcion ">")
prueba_t_par = sps.ttest_rel(t_par_df['ANTES'], t_par_df['DESPUES'], alternative='greater')
prueba_t_par
```

```python
#___ Graficar
promedio_antes = t_par_df['ANTES'].mean()
destd_antes = t_par_df['ANTES'].std()
promedio_despues = t_par_df['DESPUES'].mean()
destd_despues = t_par_df['DESPUES'].std()

plt.figure(figsize=(2,3))
plt.bar(['ANTES','DESPUES'], [promedio_antes, promedio_despues], width=0.5)
plt.errorbar(['ANTES','DESPUES'],  [promedio_antes, promedio_despues], yerr=[destd_antes,destd_despues], fmt='o', color='r', capsize=10)
plt.xlabel('MEDICION')
plt.ylabel('PRESION ARTERIAL (mmHg)')
plt.ylim(100,150)
plt.title('FUNCIONAMIENTO DE UN MEDICAMENTO')
```
